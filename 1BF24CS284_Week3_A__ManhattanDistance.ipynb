```python
import heapq

initial_state = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

goal_state = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)


def display(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


# Manhattan Distance Heuristic
def manhattan_distance(state):

    distance = 0

    for i in range(9):

        tile = state[i]

        # Ignore blank
        if tile == 0:
            continue

        # Current position
        current_row = i // 3
        current_col = i % 3

        # Goal position
        goal_index = goal_state.index(tile)

        goal_row = goal_index // 3
        goal_col = goal_index % 3

        # Calculate distance
        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# Generate possible moves
def get_neighbors(state):

    neighbors = []

    blank_index = state.index(0)

    row = blank_index // 3
    col = blank_index % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_index = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank_index], new_state[new_index] = \
                new_state[new_index], new_state[blank_index]

            neighbors.append(tuple(new_state))

    return neighbors


# A* Search using Manhattan Distance
def a_star_manhattan():

    priority_queue = []

    h = manhattan_distance(initial_state)

    heapq.heappush(
        priority_queue,
        (h, 0, initial_state)
    )

    g_cost = {
        initial_state: 0
    }

    parent = {
        initial_state: None
    }

    while priority_queue:

        f, g, current = heapq.heappop(priority_queue)

        # Goal test
        if current == goal_state:

            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            return path

        # Generate neighbors
        for neighbor in get_neighbors(current):

            new_g = g + 1

            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g
                parent[neighbor] = current

                h = manhattan_distance(neighbor)

                f = new_g + h

                heapq.heappush(
                    priority_queue,
                    (f, new_g, neighbor)
                )

    return None


# Main Program

print("A* SEARCH USING MANHATTAN DISTANCE\n")

print("Initial State:")
display(initial_state)

print("Goal State:")
display(goal_state)

print("Initial h(n) =", manhattan_distance(initial_state))

path = a_star_manhattan()

if path:

    print("Solution found!")

    print("Number of moves =", len(path) - 1)

    print("\nSolution Path:\n")

    for i, state in enumerate(path):

        print("Step", i)

        display(state)

else:

    print("No solution found.")
```
