```python
HUMAN = 'O'
COMPUTER = 'X'
EMPTY = ' '

board = [EMPTY] * 9


def print_board():
    """Print the board in a proper 3x3 format."""
    print("\n")
    print("     |     |     ")
    print(f"  {board[0]}  |  {board[1]}  |  {board[2]}  ")
    print("_____|_____|_____")
    print("     |     |     ")
    print(f"  {board[3]}  |  {board[4]}  |  {board[5]}  ")
    print("_____|_____|_____")
    print("     |     |     ")
    print(f"  {board[6]}  |  {board[7]}  |  {board[8]}  ")
    print("     |     |     ")
    print("\n")


def check_winner(player):
    """Check whether the given player has won."""
    winning_combinations = [
        (0, 1, 2),  # Row 1
        (3, 4, 5),  # Row 2
        (6, 7, 8),  # Row 3
        (0, 3, 6),  # Column 1
        (1, 4, 7),  # Column 2
        (2, 5, 8),  # Column 3
        (0, 4, 8),  # Diagonal
        (2, 4, 6)   # Diagonal
    ]

    for a, b, c in winning_combinations:
        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False


def is_draw():
    """Return True if the board is full."""
    return EMPTY not in board


def minimax(is_maximizing):
    """
    Minimax algorithm.
    Computer (X) tries to maximize the score.
    Human (O) tries to minimize the score.
    """

    # Terminal states
    if check_winner(COMPUTER):
        return 1

    if check_winner(HUMAN):
        return -1

    if is_draw():
        return 0

    if is_maximizing:
        best_score = -float('inf')

        for i in range(9):
            if board[i] == EMPTY:
                board[i] = COMPUTER
                score = minimax(False)
                board[i] = EMPTY

                best_score = max(best_score, score)

        return best_score

    else:
        best_score = float('inf')

        for i in range(9):
            if board[i] == EMPTY:
                board[i] = HUMAN
                score = minimax(True)
                board[i] = EMPTY

                best_score = min(best_score, score)

        return best_score


def computer_move():
    """Find the best move for the computer using Minimax."""
    best_score = -float('inf')
    best_move = None

    for i in range(9):
        if board[i] == EMPTY:
            board[i] = COMPUTER
            score = minimax(False)
            board[i] = EMPTY

            if score > best_score:
                best_score = score
                best_move = i

    board[best_move] = COMPUTER


def human_move():
    """Get a valid move from the human player."""
    while True:
        try:
            move = int(input("Enter your position (1-9): "))

            if move < 1 or move > 9:
                print("Please enter a number between 1 and 9.")
            elif board[move - 1] != EMPTY:
                print("That position is already occupied.")
            else:
                board[move - 1] = HUMAN
                break

        except ValueError:
            print("Please enter a valid number.")


def main():
    print("================================")
    print("       TIC-TAC-TOE GAME")
    print("================================")
    print("You are O.")
    print("Computer is X.")

    # Human starts
    while True:
        print_board()

        # Human turn
        human_move()

        if check_winner(HUMAN):
            print_board()
            print("Congratulations! You won!")
            break

        if is_draw():
            print_board()
            print("The game is a draw!")
            break

        # Computer turn
        computer_move()

        if check_winner(COMPUTER):
            print_board()
            print("Computer wins!")
            break

        if is_draw():
            print_board()
            print("The game is a draw!")
            break


if __name__ == "__main__":
    main()
```

```python
def vacuum_cleaner():
    room_A = input("Is Room A dirty? (yes/no): ").lower()
    room_B = input("Is Room B dirty? (yes/no): ").lower()
    position = input("Enter vacuum position (A/B): ").upper()

    print("\nInitial State:")
    print("Room A:", room_A)
    print("Room B:", room_B)
    print("Vacuum Position:", position)

    while room_A == "yes" or room_B == "yes":

        # Vacuum is in Room A
        if position == "A":

            if room_A == "yes":
                print("\nAction: CLEAN")
                room_A = "no"
                print("Room A cleaned.")

            elif room_B == "yes":
                print("\nAction: RIGHT")
                position = "B"
                print("Vacuum moved to Room B.")

        # Vacuum is in Room B
        elif position == "B":

            if room_B == "yes":
                print("\nAction: CLEAN")
                room_B = "no"
                print("Room B cleaned.")

            elif room_A == "yes":
                print("\nAction: LEFT")
                position = "A"
                print("Vacuum moved to Room A.")

    print("\nFinal State:")
    print("Room A: Clean")
    print("Room B: Clean")
    print("Vacuum Position:", position)
    print("\nGoal achieved! Both rooms are clean.")


vacuum_cleaner()
```
